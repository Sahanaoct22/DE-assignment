```python
# Question 9: Count frequency of each character in string "hello"
print("\nQuestion 9: Count frequency of each character in string 'hello'")
# Your code here
string = 'hello'
frequency = {}

for ch in string:
  if ch in frequency:
    frequency[ch] += 1
  else:
    frequency[ch] = 1

print(frequency)
```
